```python
   # Pandas DataFrame
# Create the transactions DataFrame
transactions = pd.DataFrame({"TransactionID":[101,102,103,104,105,106,107,108,109,110],
                             "ProductCategory":["Electronics","Clothing","Electronics","Furniture",
                                                "Clothing","Electronics","Furniture","Clothing","Furniture",
                                                "Electronics"],
                             "Region":["North","South","North","East","West","North","East","West","South","North"],
                             "Amount":[200,150,300,450,200,250,300,180,350,400]})
print(transactions)
```

```python
   # Data Exploration
# Head
transactions.head()
```

```python
# Tail
transactions.tail()
```

```python
# Shape
transactions.shape
```

```python
# Column names
transactions.columns
```

```python
# Data Type
transactions.dtypes
```

```python
# Display ProductCategory & Amount columns
transactions[["ProductCategory","Amount"]]
```

```python
# iloc
transactions.iloc[:,-3:]
```

```python
# Filter
transactions[(transactions["Region"]=="North")&(transactions["Amount"]>200)]
```

```python
# Count
transactions["ProductCategory"].value_counts()
```

```python
# Unique Value
transactions["Region"].unique()
```

```python
# Mean
transactions.groupby("Region")["Amount"].mean()
```

```python
   # Manipulating the DataFrame
# Modify Amt
transactions.loc[1,"Amount"] =165
print(transactions)
```

```python
# Discount
transactions["Discount"] = transactions["Amount"]*0.10
print(transactions)
```

```python
# Remove
transactions.drop(transactions[transactions["TransactionID"]==109].index)
```

```python
# Delete
transactions.drop("Discount",axis=1)
```
